**Exercise 4.14** Develop a discrete-time mathematical model of two species competing for the same resource, and simulate its behavior.
___

- $K$ : carrying capacity
- $x,y$ : two different predator
- $r_x, r_y$ : growth rate 
- $a$ : effect of y to x
- $b$ : effect of x to y

$$x_t = x_{t-1} + r_x x_{t-1} \left( 1- \frac{x_{t-1} + a y_{t-1}}{K_x} \right)$$
$$y_t = y_{t-1} + r_y y_{t-1} \left( 1- \frac{y_{t-1} + b x_{t-1}}{K_y} \right)$$


### Gomez' Code

```python
import matplotlib.pyplot as plt

# Initial conditions and simulation length
steps = 100
x0, y0 = 1.0, 1.0

def run_simulation(a, b, r_x, r_y, K_x, K_y):
    x, y = x0, y0
    xresult, yresult = [x], [y]
    for _ in range(steps):
        nextx = x + r_x * x * (1 - (x + a * y) / K_x)
        nexty = y + r_y * y * (1 - (y + b * x) / K_y)
        x, y = nextx, nexty
        xresult.append(x)
        yresult.append(y)
    return xresult, yresult

fig, ax = plt.subplots(figsize=(10, 6))

xres, yres = run_simulation(a=1.1, b=1.7, r_x=2.3, r_y=1.2, K_x=2.6, K_y=1.6)
(line_x,) = ax.plot(xres, 'b-', label='Species x')
(line_y,) = ax.plot(yres, 'g--', label='Species y')
ax.set_title('Two Prey Model')
ax.set_xlabel('Time')
ax.set_ylabel('Population')
ax.legend()
ax.grid(True)
```

### Cruzado's Code

```python
import matplotlib.pyplot as plt

x_pop = [40]
y_pop = [70]
rx = 0.25
ry = 0.20
Kx = 8000
Ky = 10000
a = 0.8
b = 0.6


def growth(steps=100):
    for i in range(steps):
      x_curr = x_pop[-1]
      y_curr = y_pop[-1]
      
      x_next = x_curr + rx * x_curr * (1 - (x_curr + a * y_curr) / Kx)
      y_next = y_curr + ry * y_curr * (1 - (y_curr + b * x_curr) / Ky)

      x_next = max(0, x_next)
      y_next = max(0, y_next)

      x_pop.append(x_next)
      y_pop.append(y_next)

growth(200)

plt.plot(range(len(x_pop)), x_pop, label="Species X", color="r", linewidth=2)
plt.plot(range(len(y_pop)), y_pop, label="Species Y", color="b", linewidth=2)
plt.axhline(0, color="k", linewidth=1)
plt.xlabel("Time ($t$)")
plt.ylabel("Population")
plt.title("Discrete-Time Competition Model for two Species")
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.show()
```
